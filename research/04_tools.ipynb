```python
from doctalk.config.configuration import ConfigurationManager
from doctalk.pipelines.stage_01_ingestion import IngestionPipeline
from doctalk.components.tools.summarize_tool import build_summarize_tool

# get the chunks and build the tool bound to them
chunks = IngestionPipeline().run(Path("eval/test_document.pdf"))
cm = ConfigurationManager()
summarize_tool = build_summarize_tool(cm.get_summarize_tool_config(), chunks)

print("tool name:", summarize_tool.name)
print("tool description:", summarize_tool.description)
```

```python
# call it. no arguments - the document is captured inside the tool.
summary = summarize_tool.invoke({})
print(summary)
```
