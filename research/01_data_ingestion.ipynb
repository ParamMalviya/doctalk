```python
from doctalk.config.configuration import ConfigurationManager
from doctalk.components.data_ingestion import DataIngestion

config = ConfigurationManager().get_data_ingestion_config()
ingestion = DataIngestion(config=config)

config
```

```python
chunks = ingestion.run(Path("eval/test_document.pdf"))

len(chunks)
```

```python
print("total chunks:", len(chunks))
print("metadata:", chunks[0].metadata)
print("length:", len(chunks[0].page_content))
print("---")
print(chunks[0].page_content)
```

```python
# try a different chunk size without touching config.yaml
from doctalk.entity import DataIngestionConfig

test_config = DataIngestionConfig(
    root_dir=config.root_dir,
    chunk_size=500,
    chunk_overlap=100,
)
test_chunks = DataIngestion(config=test_config).run(Path("eval/test_document.pdf"))

print(len(test_chunks), "chunks at size 500")
print(test_chunks[0].page_content)
```

```python
# which pages made it through? paper is 15 pages
pages = sorted({c.metadata["page"] for c in chunks})
print("pages present:", pages)
print("count:", len(pages))

# look at a body chunk, not the title page
body = [c for c in chunks if c.metadata["page"] in (3, 4, 5)]
print("---")
print(body[0].metadata)
print(body[0].page_content)
```
