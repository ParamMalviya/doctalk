```python
# Cell 1 - setup
import os
from pathlib import Path
if Path.cwd().name == "research":
    os.chdir("..")
from doctalk.logger import setup_logging
setup_logging()
%load_ext autoreload
%autoreload 2
from dotenv import load_dotenv
load_dotenv()
```

```python
# Cell 2 - ingest a document and build its session store
from doctalk.config.configuration import ConfigurationManager
from doctalk.pipelines.stage_01_ingestion import IngestionPipeline
from doctalk.components.vector_store import VectorStore

chunks = IngestionPipeline().run(Path("eval/test_document.pdf"))
VectorStore(ConfigurationManager().get_vector_store_config()).build(chunks, session_id="demo")
```

```python
# Cell 3 - build the chat pipeline (agent + 4 tools) for that session
from doctalk.pipelines.stage_02_chat import ChatPipeline

chat = ChatPipeline(session_id="demo", chunks=chunks)
```

```python
# Cell 4 - document question -> routes to retrieval, cites a page
print(chat.ask("How many layers does the encoder have?"))
```

```python
# Cell 5 - github question -> routes to the github tool
print(chat.ask("What's the latest commit on https://github.com/pallets/flask ?"))
```
